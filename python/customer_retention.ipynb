======
STEP 2A — Numerical Anomaly Detection
========

```python
# ==========================================
# STEP 2A — IQR OUTLIER DETECTION
# ==========================================

numerical_cols = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

for col in numerical_cols:

    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[col] < lower_bound) |
        (df_clean[col] > upper_bound)
    ]

    print(f"\n========== {col} ==========")

    print(f"Q1: {Q1:.2f}")
    print(f"Q3: {Q3:.2f}")
    print(f"IQR: {IQR:.2f}")

    print(f"Lower Bound: {lower_bound:.2f}")
    print(f"Upper Bound: {upper_bound:.2f}")

    print(f"Number of outliers: {len(outliers)}")
    print(f"Percentage of customers: {len(outliers) / len(df_clean) * 100:.2f}%")
```


### STEP 2B — Relationship-Based Anomaly Detection


```python
# ==========================================
# STEP 2B — RELATIONSHIP-BASED ANOMALY CHECK
# ==========================================

df_anomaly = df_clean.copy()

# Approximate expected charges
df_anomaly["ExpectedCharges"] = (
    df_anomaly["MonthlyCharges"] *
    df_anomaly["tenure"]
)

# Difference between actual and approximate expected charges
df_anomaly["ChargeDifference"] = (
    df_anomaly["TotalCharges"] -
    df_anomaly["ExpectedCharges"]
)

# Percentage difference
df_anomaly["ChargeDifferencePct"] = (
    df_anomaly["ChargeDifference"] /
    df_anomaly["ExpectedCharges"].replace(0, np.nan)
) * 100

print("========== CHARGE RELATIONSHIP ==========")

print(
    df_anomaly[
        [
            "tenure",
            "MonthlyCharges",
            "TotalCharges",
            "ExpectedCharges",
            "ChargeDifference",
            "ChargeDifferencePct"
        ]
    ].describe().T
)
```

```python
# ==========================================
# CUSTOMERS WITH LARGEST CHARGE DIFFERENCES
# ==========================================

print("========== TOP 20 CHARGE DIFFERENCES ==========")

top_charge_diff = df_anomaly[
    df_anomaly["ExpectedCharges"] > 0
].sort_values(
    "ChargeDifferencePct",
    ascending=False
)

print(
    top_charge_diff[
        [
            "customerID",
            "tenure",
            "MonthlyCharges",
            "TotalCharges",
            "ExpectedCharges",
            "ChargeDifference",
            "ChargeDifferencePct",
            "Contract",
            "InternetService",
            "Churn"
        ]
    ].head(20)
)
```

```python
print("========== LOWEST CHARGE DIFFERENCES ==========")

bottom_charge_diff = df_anomaly[
    df_anomaly["ExpectedCharges"] > 0
].sort_values(
    "ChargeDifferencePct",
    ascending=True
)

print(
    bottom_charge_diff[
        [
            "customerID",
            "tenure",
            "MonthlyCharges",
            "TotalCharges",
            "ExpectedCharges",
            "ChargeDifference",
            "ChargeDifferencePct",
            "Contract",
            "InternetService",
            "Churn"
        ]
    ].head(20)
)
```

### STEP 2C — High-Value Churn Anomaly

```python
# ==========================================
# STEP 2C — HIGH-VALUE CHURN CUSTOMERS
# ==========================================

# Find the 75th percentile of MonthlyCharges
high_value_threshold = df_clean["MonthlyCharges"].quantile(0.75)

# Find the 75th percentile of tenure
long_tenure_threshold = df_clean["tenure"].quantile(0.75)

high_value_churn = df_clean[
    (df_clean["MonthlyCharges"] >= high_value_threshold) &
    (df_clean["tenure"] >= long_tenure_threshold) &
    (df_clean["Churn"] == "Yes")
].copy()

print("========== HIGH-VALUE CHURN CUSTOMERS ==========")

print("High monthly charge threshold:",
      round(high_value_threshold, 2))

print("Long tenure threshold:",
      long_tenure_threshold)

print("Number of high-value churn customers:",
      len(high_value_churn))

print("\nPotential monthly revenue lost:")

print(
    round(high_value_churn["MonthlyCharges"].sum(), 2)
)

print("\nTop high-value churn customers:")

print(
    high_value_churn[
        [
            "customerID",
            "tenure",
            "MonthlyCharges",
            "TotalCharges",
            "Contract",
            "InternetService",
            "PaymentMethod",
            "Churn"
        ]
    ]
    .sort_values("MonthlyCharges", ascending=False)
    .head(20)
)
```

```python
# ==========================================
# EARLY-TENURE CHURN
# ==========================================

early_churn = df_clean[
    (df_clean["tenure"] <= 6) &
    (df_clean["Churn"] == "Yes")
].copy()

print("========== EARLY-TENURE CHURN ==========")

print("Early churn customers:",
      len(early_churn))

print("Percentage of all churn:",
      round(
          len(early_churn) /
          len(df_clean[df_clean["Churn"] == "Yes"]) * 100,
          2
      ),
      "%")

print("\nAverage monthly charge of early churn customers:",
      round(early_churn["MonthlyCharges"].mean(), 2))
```

## STEP 3 — Exploratory Data Analysis

#### Step 3A — Start with the most basic business KPI

```python
# ==========================================
# STEP 3A — CORE BUSINESS KPIs
# ==========================================

total_customers = df_clean["customerID"].nunique()

churned_customers = (
    df_clean[df_clean["Churn"] == "Yes"]["customerID"].nunique()
)

active_customers = (
    df_clean[df_clean["Churn"] == "No"]["customerID"].nunique()
)

churn_rate = churned_customers / total_customers * 100

monthly_revenue = df_clean["MonthlyCharges"].sum()

monthly_revenue_lost = (
    df_clean.loc[
        df_clean["Churn"] == "Yes",
        "MonthlyCharges"
    ].sum()
)

annual_revenue_lost = monthly_revenue_lost * 12

print("========== CORE BUSINESS KPIs ==========")

print(f"Total customers: {total_customers:,}")

print(f"Active customers: {active_customers:,}")

print(f"Churned customers: {churned_customers:,}")

print(f"Churn rate: {churn_rate:.2f}%")

print(f"Total monthly revenue: {monthly_revenue:,.2f}")

print(f"Monthly revenue from churned customers: "
      f"{monthly_revenue_lost:,.2f}")

print(f"Annualized revenue exposure: "
      f"{annual_revenue_lost:,.2f}")
```

#### Step 3B — Churn by Contract Type

```python
# ==========================================
# STEP 3B — CHURN BY CONTRACT
# ==========================================

contract_analysis = (
    df_clean
    .groupby("Contract")
    .agg(
        TotalCustomers=("customerID", "nunique"),
        ChurnedCustomers=("Churn", lambda x: (x == "Yes").sum()),
        MonthlyRevenue=("MonthlyCharges", "sum")
    )
    .reset_index()
)

contract_analysis["ChurnRate"] = (
    contract_analysis["ChurnedCustomers"] /
    contract_analysis["TotalCustomers"] * 100
)

contract_analysis["RevenueFromChurnedCustomers"] = (
    df_clean[df_clean["Churn"] == "Yes"]
    .groupby("Contract")["MonthlyCharges"]
    .sum()
    .reindex(contract_analysis["Contract"])
    .fillna(0)
    .values
)

contract_analysis = contract_analysis.sort_values(
    "ChurnRate",
    ascending=False
)

print(contract_analysis)
```

#### Step 3I — Revenue / Monthly Charges

```python
df_clean["MonthlyChargeGroup"] = pd.qcut(
    df_clean["MonthlyCharges"],
    q=4,
    labels=[
        "Low",
        "Medium-Low",
        "Medium-High",
        "High"
    ]
)

charge_analysis = (
    df_clean
    .groupby("MonthlyChargeGroup", observed=True)
    .agg(
        TotalCustomers=("customerID", "nunique"),
        ChurnedCustomers=("Churn", lambda x: (x == "Yes").sum()),
        AvgMonthlyCharges=("MonthlyCharges", "mean"),
        MonthlyRevenue=("MonthlyCharges", "sum")
    )
    .reset_index()
)

charge_analysis["ChurnRate"] = (
    charge_analysis["ChurnedCustomers"] /
    charge_analysis["TotalCustomers"] * 100
)

print(charge_analysis)
```

#### Step 3K — Find the Highest-Risk Customer Segments

```python
risk_segment = (
    df_clean
    .groupby(
        ["Contract", "InternetService", "PaymentMethod"]
    )
    .agg(
        TotalCustomers=("customerID", "nunique"),
        ChurnedCustomers=("Churn", lambda x: (x == "Yes").sum()),
        MonthlyRevenue=("MonthlyCharges", "sum"),
        ChurnedMonthlyRevenue=(
            "MonthlyCharges",
            lambda x: x[df_clean.loc[x.index, "Churn"] == "Yes"].sum()
        )
    )
    .reset_index()
)

risk_segment["ChurnRate"] = (
    risk_segment["ChurnedCustomers"] /
    risk_segment["TotalCustomers"] * 100
)

risk_segment = risk_segment.sort_values(
    "ChurnRate",
    ascending=False
)

print(risk_segment.head(15))
```

#### Step 3L — Final EDA: Churn Driver Summary

```python
churn_summary = []

features = [
    "Contract",
    "PaymentMethod",
    "InternetService",
    "OnlineSecurity",
    "TechSupport",
    "OnlineBackup",
    "DeviceProtection",
    "StreamingTV",
    "StreamingMovies"
]

for feature in features:

    temp = (
        df_clean
        .groupby(feature)
        .agg(
            TotalCustomers=("customerID", "nunique"),
            ChurnedCustomers=("Churn", lambda x: (x == "Yes").sum())
        )
        .reset_index()
    )

    temp["ChurnRate"] = (
        temp["ChurnedCustomers"] /
        temp["TotalCustomers"] * 100
    )

    temp["Feature"] = feature
    temp["Segment"] = temp[feature].astype(str)

    churn_summary.append(
        temp[
            [
                "Feature",
                "Segment",
                "TotalCustomers",
                "ChurnedCustomers",
                "ChurnRate"
            ]
        ]
    )

churn_summary = pd.concat(
    churn_summary,
    ignore_index=True
)

churn_summary = churn_summary.sort_values(
    "ChurnRate",
    ascending=False
)

print(churn_summary.head(25))
```
